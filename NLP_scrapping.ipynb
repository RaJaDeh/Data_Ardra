```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
import time
import pandas as pd

options = Options()
options.add_argument("--headless")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36")
driver = webdriver.Chrome(options=options)

base_url = "https://www.flipkart.com/redmi-6-black-64-gb/product-reviews/itm3b52cae7d753e?pid=MOBFA2NXDQFTXNHJ"

reviews, ratings = [], []

for page in range(1, 100):  # scrape 10 halaman
    url = base_url + f"&page={page}"
    driver.get(url)
    time.sleep(2)

    review_blocks = driver.find_elements(By.CLASS_NAME, "col.EPCmJX.Ma1fCG")
    if not review_blocks:
        print(f"❌ Halaman {page} kosong atau diblok!")
        continue

    for block in review_blocks:
        try:
            rating = block.find_element(By.CLASS_NAME, "XQDdHH").text.strip()
            comment = block.find_element(By.CLASS_NAME, "ZmyHeo").text.strip()
            ratings.append(rating)
            reviews.append(comment)
        except:
            continue

driver.quit()

df = pd.DataFrame({'rating': ratings, 'review': reviews})
df.to_csv("flipkart_redmi6_reviews_with_rating.csv", index=False, encoding='utf-8-sig')
print("✅ Selesai! Total review terkumpul:", len(df))
```

```python
import pandas as pd
from transformers import pipeline
```

```python
model = pipeline("sentiment-analysis", model="nlptown/bert-base-multilingual-uncased-sentiment")

# STEP 6: Prediksi sentimen dari kolom 'gabungan'
def map_label(label):
    rating = int(label[0])
    if rating >= 4:
        return 'positif'
    elif rating == 3:
        return 'netral'
    else:
        return 'negatif'

results = model(df['review'].astype(str).tolist(), truncation=True)

df['label_help'] = [map_label(r['label']) for r in results]
```

```python
df['rating'] = df['rating'].astype(float)
```
